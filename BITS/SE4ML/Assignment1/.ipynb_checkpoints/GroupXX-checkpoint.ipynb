## Domain & Problem Statement

**Domain:** Financial Services - real-time payment fraud detection.

**Problem Statement:** Banks and payment processors must decide, within a couple of hundred milliseconds, whether an incoming credit-card transaction is legitimate or fraudulent - *before* the transaction is authorized at the point of sale. Static, rule-based fraud filters are easy for fraudsters to reverse-engineer, drift out of date, and tend to generate a high rate of false declines that damage genuine customers' experience. This project designs, models (using GR4ML) and implements a **real-time, ML-based fraud detection system** that:

1. scores every transaction for fraud probability at the moment of authorization,
2. is periodically retrained offline on newly labelled transaction data as fraud patterns evolve, and
3. keeps ML scoring and business decision policy (thresholds, blacklists, compliance overrides) in separate, independently-changeable components.

This notebook covers both assignment objectives:
- **Objective 1** - Requirements formulation using GR4ML notation (Business View, Analytics Design View, Data Preparation View) and the top-3 quality requirements.
- **Objective 2** - System architecture (ML + non-ML components) implementing two architectural patterns: a **Lambda Architecture** and a **Microservices Architecture**.

## Objective 1: Requirements Formulation

We use **GR4ML** (Goal-oriented Requirements engineering for Machine Learning; Nalchigar & Yu, University of Toronto) to move from business goals to a concrete ML solution design. GR4ML uses three linked, goal-oriented views:

| View | Question it answers | Key elements |
|---|---|---|
| **Business View** | *Why?* Who needs ML, for what business goal | Actor, Strategic Goal, Indicator, Decision Goal, Question Goal, Insight |
| **Analytics Design View** | *What?* What analytics/ML approach is needed | Analytics Goal (Prediction / Description / Prescription), Algorithm, Softgoal |
| **Data Preparation View** | *How?* How is data readied for the analytics | Entity, Data Preparation Task (Cleaning / Integration / Transformation / Reduction), Prepared Dataset |

The three views are linked end-to-end: a Business-View **Insight** *generates* an Analytics-View **Analytics Goal**, which in turn *requires* a Data-Preparation-View **Prepared Dataset**.


### 1.1 Business View

The diagram below elicits *why* the bank needs this ML capability: the Fraud & Risk Ops team wants to minimize fraud losses **and** maintain customer trust (these two strategic goals are in natural tension, which is what makes this a hard decision problem, not just a classification exercise). Each strategic goal is measured by concrete indicators. The **Decision Goal** - approve / flag / decline a transaction in real time - is what actually needs ML support, because a human cannot review millions of transactions individually. That decision goal raises two **Question Goals**, each answered by an **Insight** that becomes the input to the Analytics Design View.

### 1.3 Data Preparation View

The Prediction Goal *requires* a prepared dataset. Raw data lives in three entities - **Transactions**, **Customers**, **Merchants** - and is turned into an 8-feature model-ready table through four Data Preparation Tasks: **Cleaning** (de-duplication, missing-value imputation), **Integration** (joining transactions with customer/merchant profiles), **Transformation** (deriving behavioural signals: amount-to-average ratio, geo-distance from home, night-time flag, transaction-velocity flag) and **Reduction** (selecting the final 8 features and applying class-weighting to handle the ~1.5% fraud rate).

Critically, this pipeline (`src/features.py`) is shared *verbatim* between the offline batch-training code and the online `feature_service` microservice, so training/serving feature skew cannot occur.

### 1.4 Top Three Quality Requirements (Softgoals) and Justification

Based on the influence links in the Analytics Design View and the Indicators in the Business View, we prioritize the following three quality requirements:

1. **Low latency (real-time scoring, target < 200 ms end-to-end).**
   *Justification:* the Decision Goal sits directly in the payment-authorization path (Business View Indicator: *Approval latency*). If scoring is slow, the transaction either times out (lost sale, bad customer experience) or the bank is forced to approve without a score (defeats the purpose). This is a hard real-time constraint, not just a "nice to have".

2. **High recall with a controlled false-positive rate, under extreme class imbalance (~1.5% fraud).**
   *Justification:* directly reflects the tension between the two Strategic Goals - "minimize fraud losses" (needs high recall, i.e. catch as much fraud as possible) versus "maintain customer trust" (needs high precision, i.e. do not decline genuine customers). Because fraud is rare, a naive accuracy-optimizing model would predict "legitimate" for everything and still be >98% "accurate" while catching zero fraud - so recall/precision (not accuracy) must be the explicit, monitored quality target.

3. **Scalability and availability at peak transaction volume.**
   *Justification:* the system must keep scoring transactions during traffic spikes (festive-season sales, salary day) without degrading latency or dropping requests, since an outage in the scoring path directly blocks revenue-generating transactions. This requirement is what motivates decomposing the system into independently-scalable microservices (Section 2.2, Pattern 2) rather than a monolith.

*(A fourth quality requirement, interpretability/auditability of individual fraud decisions, is important for regulatory compliance (RBI/PCI-DSS) but was not put in the top three because it can be adequately addressed post-hoc via the Random Forest's built-in feature importances and the fact that business rules are kept in an auditable, human-readable Decision Service rather than the model itself.)*

### 2.2 Architectural Patterns Selected

**Pattern 1 - Lambda Architecture.**
The system needs two different processing speeds: (a) a **batch layer** that periodically retrains the fraud model on the full historical transaction log (accuracy over speed - can take minutes), and (b) a **speed layer** that scores each incoming transaction the instant it arrives (speed over completeness - must respond in milliseconds using whatever model is currently published). The batch layer publishes versioned model artifacts to a model registry; the speed layer always uses the latest published model. This is a textbook fit for Lambda Architecture, and it cleanly separates "when the model gets smarter" from "how fast a single transaction gets scored".

**Pattern 2 - Microservices Architecture.**
The speed layer itself is decomposed into independently deployable services - **Ingestion**, **Feature**, **Inference**, **Decision**, **Notification** - each with its own REST API. This matters for this domain specifically because:
- the **Inference Service** (the only ML component in the request path) can be scaled horizontally / given a GPU / swapped to a different framework, independently of the other services, when load increases;
- the **Decision Service** encodes business/compliance rules that change frequently (new blacklists, new thresholds) - it can be redeployed without touching the ML model or retraining anything;
- a failure in the **Notification Service** (e.g. an alerting backend is down) does not block the approve/decline decision itself, improving fault isolation and availability - one of our top-3 quality requirements.


### 2.3 Implementation

#### 2.3.1 Pattern 1 - Lambda Architecture: Batch Layer (model training)

The batch layer trains a class-weighted Random Forest on the prepared dataset from Section 1.3, evaluates it with metrics appropriate for extreme class imbalance (ROC-AUC, PR-AUC, precision/recall/F1 on the fraud class - **not** plain accuracy), and publishes the model artifact that the speed layer / inference microservice will load.


```python
from train_batch import run_batch_training

model, metrics, (X_test, y_test, proba, cm) = run_batch_training(n_transactions=60_000)
metrics
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import precision_recall_curve

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["legit", "fraud"],
            yticklabels=["legit", "fraud"], ax=axes[0], cbar=False)
axes[0].set_xlabel("Predicted"); axes[0].set_ylabel("Actual"); axes[0].set_title("Confusion Matrix (threshold=0.5)")

prec, rec, thr = precision_recall_curve(y_test, proba)
axes[1].plot(rec, prec, color="#1565C0")
axes[1].set_xlabel("Recall"); axes[1].set_ylabel("Precision")
axes[1].set_title(f"Precision-Recall Curve (PR-AUC={metrics['pr_auc']:.3f})")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(ARTIFACTS / "batch_eval_plots.png", dpi=150, bbox_inches="tight")
plt.show()
```

The precision-recall curve makes the recall-vs-precision quality trade-off (Section 1.4, requirement #2) concrete and tunable: the Decision Service's threshold (Section 2.3.2) can be moved along this curve to favor recall (catch more fraud, tolerate more false declines) or precision (fewer false declines, risk missing some fraud) *without retraining the model* - exactly the separation of concerns the microservices pattern is designed to give us.

#### 2.3.2 Pattern 2 - Microservices Architecture: implementation

Each service below is a small, independently runnable **FastAPI** application (`src/services/*.py`), each with its own Swagger UI (`/docs`). The **Inference Service** is the only one that touches the ML model - it simply loads the artifact published by the batch layer above.

| Service | Port | Type | Responsibility |
|---|---|---|---|
| `ingestion_service` | 8001 | non-ML | Public API gateway; validates the raw transaction and orchestrates the pipeline below over HTTP |
| `feature_service` | 8002 | non-ML | Looks up customer/merchant profiles and computes the 8-feature vector (same code as Section 1.3) |
| `inference_service` | 8003 | **ML** | Loads `artifacts/model.pkl` and returns a fraud probability |
| `decision_service` | 8004 | non-ML | Applies business rules (threshold, merchant blacklist) to turn a score into approve/flag/decline |
| `notification_service` | 8005 | non-ML | Raises alerts for flagged/declined transactions |

The cell below starts all five as **separate OS processes** (`uvicorn`, one per service/port) and fires real transactions at the public `ingestion_service` endpoint over HTTP - this is the same code path a live payment gateway would use. **While this cell is running**, you can open `http://127.0.0.1:8001/docs` (and 8002-8005) in a browser to see each service's live Swagger UI - screenshot those pages for the assignment report as the "application screenshots" evidence.

### 2.4 How the two patterns fit together

The Lambda Architecture's **speed layer** is not a separate piece of code - it *is* the `feature_service -> inference_service -> decision_service` chain of the Microservices Architecture, called synchronously per transaction. The **batch layer** (offline, scheduled) and the **speed layer** (online, per-request) share nothing except the model artifact and the feature-engineering code, which is exactly the coupling Lambda Architecture prescribes: independent freshness/latency characteristics, one shared contract.
